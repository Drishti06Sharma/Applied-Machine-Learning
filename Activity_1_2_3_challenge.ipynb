## 3. Description

Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value. It assumes that there is a linear relationship between the input variable and the output variable.

### Simple Linear Regression

For Simple Linear Regression:

**y = mx + c**

Where:

- **x** = Independent / input variable
- **y** = Dependent / output variable
- **m** = Slope / coefficient
- **c** = Intercept

In machine learning notation:

**ŷ = β₀ + β₁x**

### Mean Squared Error (MSE)

**MSE = (1/n) Σ(y − ŷ)²**

MSE measures the average squared difference between the actual and predicted values.

**Lower MSE indicates better performance.**

<h3>Linear Regression Using NumPy</h3>

</div>

### Problem Statement

Suppose we want to predict a student's marks based on the number of hours studied.

The regression model will be used to study the relationship between **Hours Studied** and **Marks** and predict marks for new students.

### Step 1: Create the Dataset

The following dataset contains the number of hours studied and the corresponding marks obtained by students.

```python
import numpy as np

hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([35, 40, 50, 55, 65, 70, 75, 85])
```

### Step 2: Display the Dataset

```python
print("Hours:", hours)
print("Marks:", marks)
```

### Step 3: Calculate Basic Statistics

Calculate the mean of hours, mean of marks, standard deviation of marks, minimum marks, and maximum marks.

```python
print("Mean Hours:", np.mean(hours))
print("Mean Marks:", np.mean(marks))
print("Standard Deviation:", np.std(marks))
print("Minimum Marks:", np.min(marks))
print("Maximum Marks:", np.max(marks))
```

### Step 4: Visualize the Relationship

The scatter plot shows the relationship between the number of hours studied and the marks obtained.

```python
import matplotlib.pyplot as plt

plt.scatter(hours, marks)
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours Studied vs Marks")
plt.show()
```

### Step 5: Find the Regression Equation

NumPy provides the `polyfit()` function for fitting a polynomial. For linear regression, the degree is set to **1**.

The regression equation has the form:

**ŷ = mx + c**

```python
m, c = np.polyfit(hours, marks, 1)

print("Slope:", m)
print("Intercept:", c)
```

### Step 6: Calculate Predictions

Use the regression equation to calculate the predicted marks for the existing observations.

```python
predicted_marks = m * hours + c

print("Predicted Marks:")
print(predicted_marks)
```

### Step 7: Predict Marks for a New Student

Predict the marks for a student who studies **6.5 hours**.

```python
new_hours = 6.5
new_prediction = m * new_hours + c

print("Predicted Marks for 6.5 hours:", new_prediction)
```

### Step 8: Predict Marks for Multiple Students

Predict marks for students studying **2.5, 4.5, 7.5, and 10 hours**.

```python
new_hours = np.array([2.5, 4.5, 7.5, 10])
new_predictions = m * new_hours + c

for h, p in zip(new_hours, new_predictions):
    print(f"{h} hours -> {p:.2f} marks")
```

### Step 9: Plot the Regression Line

The scatter points represent the actual observations, while the straight line represents the values predicted by the regression model.

```python
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted_marks, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression")
plt.legend()
plt.show()
```

### Step 10: Calculate Errors and Model Performance

**Error = Actual − Predicted**

The prediction errors are calculated to evaluate how closely the regression model predicts the actual marks.

```python
errors = marks - predicted_marks

print("Errors:")
print(errors)

mse = np.mean((marks - predicted_marks) ** 2)
rmse = np.sqrt(mse)

print("Mean Squared Error:", mse)
print("RMSE:", rmse)
```

<h3>Linear Regression Using Scikit-learn</h3>

</div>

### Step 1: Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

### Step 2: Define the Dataset and Reshape the Input

Scikit-learn expects the input features **X** in a two-dimensional format:

**(number of samples, number of features)**

```python
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([35, 40, 50, 55, 65, 70, 75, 85])

X = hours.reshape(-1, 1)
y = marks

print("X shape:", X.shape)
print("y shape:", y.shape)
```

### Step 3: Create and Train the Model

```python
model = LinearRegression()
model.fit(X, y)

print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

### Step 4: Predict the Existing Data

```python
predicted = model.predict(X)

print("Predicted Marks:")
print(predicted)
```

### Step 5: Predict Marks for a New Student

Predict the marks for a student who studies **6.5 hours**.

```python
new_student = np.array([[6.5]])

prediction = model.predict(new_student)

print("Predicted Marks for 6.5 hours:", prediction[0])
```

### Step 6: Evaluate the Model

The model is evaluated using:

- **MAE:** Mean Absolute Error
- **MSE:** Mean Squared Error
- **RMSE:** Root Mean Squared Error
- **R² Score:** Coefficient of determination

```python
mae = mean_absolute_error(y, predicted)
mse = mean_squared_error(y, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

### Step 7: Visualize the Regression Line

The graph shows the actual data points and the regression line fitted by the Linear Regression model.

```python
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression: Hours Studied vs Marks")
plt.legend()
plt.show()
```

<h2>COMPLETE PROGRAM</h2>

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Dataset
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([35, 40, 50, 55, 65, 70, 75, 85])

# Reshape input
X = hours.reshape(-1, 1)
y = marks

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Model parameters
print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)

# Predictions
predicted = model.predict(X)

# Evaluation metrics
mae = mean_absolute_error(y, predicted)
mse = mean_squared_error(y, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted)

print("\nModel Evaluation")
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)

# Prediction for a new student
new_student = np.array([[6.5]])
prediction = model.predict(new_student)

print("\nPrediction for 6.5 hours:", prediction[0])

# Visualization
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression: Hours Studied vs Marks")
plt.legend()
plt.show()
```

<h2>ACTIVITY #3</h2>
<h3>Linear Regression with a Modified Dataset</h3>

</div>

### Objective

Change the dataset and train the Linear Regression model again.

Calculate:

- **Coefficient**
- **Intercept**
- **MAE**
- **MSE**
- **RMSE**
- **R² Score**

Then predict the marks for **3.5, 5.5, 7.5, and 10 hours**.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Modified Dataset
hours = np.array([2, 3, 4, 5, 6, 7, 8, 9])
marks = np.array([42, 48, 55, 61, 68, 74, 81, 88])

# Reshape input
X = hours.reshape(-1, 1)
y = marks

# Train model
model = LinearRegression()
model.fit(X, y)

# Model parameters
print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)

# Predictions
predicted = model.predict(X)

# Evaluation
mae = mean_absolute_error(y, predicted)
mse = mean_squared_error(y, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)

# Multiple predictions
new_hours = np.array([3.5, 5.5, 7.5, 10]).reshape(-1, 1)
new_predictions = model.predict(new_hours)

print("\nPredictions:")
for h, p in zip(new_hours.flatten(), new_predictions):
    print(f"{h} hours -> {p:.2f} marks")

# Regression plot
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression with Modified Dataset")
plt.legend()
plt.show()
```

### Problem Statement

Consider the following dataset representing advertising expenditure and sales.

The objective is to build a Linear Regression model to predict sales from advertising expenditure.

### Tasks

1. Visualize the dataset.
2. Train a Linear Regression model.
3. Find the coefficient and intercept.
4. Plot the regression line.
5. Calculate MAE, MSE, RMSE and R² Score.
6. Predict sales when advertising expenditure is **55**.
7. Predict sales for advertising expenditures of **25, 45 and 75**.
8. Interpret the coefficient in the context of advertising and sales.

### Step 1: Define the Dataset

The following dataset represents advertising expenditure and corresponding sales.

```python
import numpy as np

advertising = np.array([10, 20, 30, 40, 50, 60, 70, 80])
sales = np.array([25, 30, 38, 45, 50, 58, 65, 72])

X = advertising.reshape(-1, 1)
y = sales

print("Advertising Expenditure:", advertising)
print("Sales:", sales)
```

### Step 2: Train the Linear Regression Model

```python
model = LinearRegression()
model.fit(X, y)

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

### Step 3: Make Predictions

The trained model is used to:

- Predict sales when advertising expenditure is **55**.
- Predict sales for advertising expenditures of **25, 45, and 75**.

```python
single_value = np.array([[55]])
single_prediction = model.predict(single_value)

print("Predicted Sales for 55:", single_prediction[0])

multiple_values = np.array([[25], [45], [75]])
multiple_predictions = model.predict(multiple_values)

print("\nPredictions:")
for x, p in zip(multiple_values.flatten(), multiple_predictions):
    print(f"Advertising Expenditure {x} -> Predicted Sales {p:.2f}")
```

### Step 4: Evaluate the Model

The model performance is evaluated using:

- **MAE**
- **MSE**
- **RMSE**
- **R² Score**

```python
predicted_sales = model.predict(X)

mae = mean_absolute_error(y, predicted_sales)
mse = mean_squared_error(y, predicted_sales)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted_sales)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

### Step 5: Visualize the Regression Line

The scatter plot represents the actual relationship between advertising expenditure and sales, while the line represents the values predicted by the regression model.

```python
plt.scatter(advertising, sales, label="Actual Data")
plt.plot(advertising, predicted_sales, label="Regression Line")
plt.xlabel("Advertising Expenditure")
plt.ylabel("Sales")
plt.title("Advertising Expenditure vs Sales")
plt.legend()
plt.show()
```

### Interpretation

The coefficient represents the expected change in sales for every one-unit increase in advertising expenditure.

A positive coefficient indicates that higher advertising expenditure is associated with higher predicted sales.
