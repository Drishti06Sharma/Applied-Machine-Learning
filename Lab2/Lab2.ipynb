# Experiment: Telco Customer Churn Dataset – Data Preprocessing

## Aim
To perform data loading, exploration, cleaning, and preprocessing on the Telco Customer Churn dataset using Python libraries.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

## 1. Loading the Dataset

The Telco Customer Churn dataset is loaded using Pandas. The first five records are displayed to understand the structure and contents of the dataset.

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
print(df.head())
```

## 2. Dataset Dimensions

The shape of the dataset is checked to determine the total number of rows and columns. The dataset contains 7043 records and 21 columns.

```python
print("Shape of dataset:", df.shape)
```

## 3. Displaying Column Names

All column names are displayed to identify the available features, customer details, service information, billing attributes, and target variable.

```python
print(df.columns)
```

## 4. Checking Missing Values

The dataset is checked for missing values using the isnull() function. Initially, no missing values are detected because some blank entries are stored as strings.

```python
print(df.isnull().sum())
```

## 5. Converting TotalCharges into Numeric Format

The TotalCharges column is converted into numeric data type using pd.to_numeric(). The errors="coerce" parameter converts invalid values into NaN, allowing missing values to be identified.

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

## 6. Handling Missing Values

Missing values are identified after converting TotalCharges into numeric format. The dropna() function is used to remove rows containing missing values, resulting in a cleaned dataset.

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
df = df.dropna()
```

## 7. Target Variable Distribution

The value_counts() function is used to examine the distribution of customers who churned and those who did not churn. This helps understand the distribution of the target variable.

```python
print(df["Churn"].value_counts())
```

## 8. Removing Irrelevant Feature

The customerID column is removed because it is a unique identifier and does not provide useful information for machine learning prediction.

```python
df = df.drop("customerID", axis=1)
```

## 9. Encoding Categorical Variables

Categorical variables are converted into numerical format using One-Hot Encoding with pd.get_dummies(). The drop_first=True parameter removes the first category of each variable to avoid redundant features.

```python
df = pd.get_dummies(df, drop_first=True)
```

The Telco Customer Churn dataset was successfully loaded, explored, and preprocessed. Missing values were handled, the customerID column was removed, and categorical variables were converted into numerical format using One-Hot Encoding. The final dataset is ready for further machine learning analysis.
