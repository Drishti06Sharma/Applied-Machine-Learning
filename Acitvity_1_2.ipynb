## Linear regression using numpy

# Dataset

```python
##dataset
import numpy as np
hours=np.array([1,2,3,4,5,6,7,8])
marks=np.array([35,40,50,55,65,70,75,85])
```

# display data

```python


print("Hours:",hours)
print("Marks:",marks)
```

# calc basic stats

```python

print("Mean of hours:",np.mean(hours))
print("Mean of marks:",np.mean(marks))
print("Standard deviation of hours:",np.std(hours))
print("Maximum of hours:",np.max(hours))
print("Minimum of hours:",np.min(hours))
```

# create a scatter plot

```python
import matplotlib.pyplot as plt

plt.scatter(hours,marks)    
plt.xlabel('Hours Studied')
plt.ylabel('Marks')
plt.title('Scatter plot of Hours vs Marks')
plt.show()
```

## Linear Regression Using NumPy

```python
m, c = np.polyfit(hours, marks, 1)
print("Slope:", m)
print("Intercept:", c)
```

## Calc predictions

```python
predicted_marks = m * hours + c
print("Predicted Marks:")
print(predicted_marks)
```

## Predict Marks for a New Student (6.5 hours)

```python
new_hours = 6.5
predicted = m * new_hours + c
print("Predicted Marks:", predicted)
```

### Predict marks for 2.5, 4.5, 7.5, and 10 hours

```python
new_hours_list = np.array([2.5, 4.5, 7.5, 10])
predictions = m * new_hours_list + c
print("Predictions for [2.5, 4.5, 7.5, 10] hours:", predictions)
```

## Plot the Regression Line

```python
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted_marks, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression")
plt.legend()
plt.show()
```

## Calculate Errors

```python
# Error = Actual - Predicted
errors = marks - predicted_marks
print("Errors:")
print(errors)

# Mean Squared Error (MSE)
mse = np.mean((marks - predicted_marks) ** 2)
print("Mean Squared Error:", mse)

# Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)
print("RMSE:", rmse)
```

## Import LinearRegression & Metrics


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## Define Dataset & Reshape Input

```python
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([35, 40, 50, 55, 65, 70, 75, 85])

# Scikit-learn expects 2D array for input features (X)
X = hours.reshape(-1, 1)
y = marks

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Create, Train Model & Display Parameters

```python
model = LinearRegression()
model.fit(X, y)

print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Predict Existing Data

```python
predicted = model.predict(X)
print("Predicted Marks for training data:")
print(predicted)
```

## Predict Marks for New Student (6.5 hours)

```python
new_student = np.array([[6.5]])
prediction = model.predict(new_student)
print("Predicted Marks for 6.5 hours:", prediction[0])
```

## Evaluate Model Metrics

```python
mae = mean_absolute_error(y, predicted)
mse = mean_squared_error(y, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

## Visualize Graph (Attractive Matplotlib View)

```python
# Visualization
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression: Hours Studied vs Marks")
plt.legend()
plt.show()
```
