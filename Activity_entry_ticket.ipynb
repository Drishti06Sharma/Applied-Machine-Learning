```python
import numpy as np
X=np.array([1,2,3,4,5])
Y=np.array([35,40,50,55,65])

m,c=np.polyfit(X,Y,1)
print('Slope:',m)
print('Intercept:',c)

y_predicted = m*X + c
error = Y - y_predicted
MAE = np.mean(np.abs(error))
print('Mean Absolute Error:', MAE)
MSE = np.mean(error**2)
print('Mean Squared Error:', MSE)
RMSE = np.sqrt(MSE)
print('Root Mean Squared Error:', RMSE) 
```
