```python
import numpy as np
x=np.array([1,2,3,4,5])
y=np.array([35,40,45,50,55,60])
m,c=np.polyfit(x,y,1)
print("Slope:",m)
print("Intercept:",c)

```

```python
import numpy as np

x = np.array(list(map(float, input("Enter number of rooms: ").split())))
y = np.array(list(map(float, input("Enter prices: ").split())))

m, c = np.polyfit(x, y, 1)

print("Slope:", m)
print("Intercept:", c)

# Prediction
rooms = float(input("Enter rooms to predict price: "))
predicted_price = m * rooms + c

print("Predicted Price:", predicted_price)
```
